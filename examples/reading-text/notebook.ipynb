# Reading text: commits and requirement sentences

Human text that follows rules, read by grammar (`resources/grammars/humanish/`). Runs on this
repository itself - no network needed. Translation is at the end and needs the `translate`
extra and an installed language package.

```python
from gitrecon.humanish import parse_commit

commit = parse_commit("feat(cli)!: add gitgraph\n\nBREAKING CHANGE: the old flag is gone\nRefs #12")
commit.form, commit.type, commit.scope, commit.work, commit.breaking, commit.trailers
```

```python
from gitrecon.config import PROJECT_ROOT
from gitrecon.humanish import commit_labels, read_commits, summarize_commits

commits = read_commits(PROJECT_ROOT)
summary = summarize_commits(commits)
print(summary["forms"], summary["conventional_share"])
print(summary["work"])
[str(label) for label in commit_labels("repo:sarverott/gitrecon", commits)]
```

## Requirement sentences

Keywords of RFC 2119 in capitals; a keyword in quotes is a mention, not a requirement.

```python
from gitrecon.humanish import find_requirements

text = '''A client MUST send the header, e.g. X-A. It MAY retry.
Servers SHOULD NOT cache the answer and MUST NOT log it. The word "MUST" is explained elsewhere.'''
[(r.level, r.keywords, r.sentence) for r in find_requirements(text)]
```

## Translation (optional)

`gitrecon translate install en pl` once, then:

```python
from gitrecon import translate

try:
    print(translate.installed_languages())
    print(translate.translate("A client MUST send the header.", "en", "pl"))
except (RuntimeError, LookupError) as error:
    print("not available here:", error)
```

CLI: `gitrecon commits [PATH]`, `gitrecon text requirements URL|FILE [--raw]`,
`gitrecon translate text|languages|install`.
