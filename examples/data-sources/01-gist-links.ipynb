# Data sources 1/5: links harvested from gists

Notes kept in gists are full of links worth watching. `harvest_gists()` reads only the
directories cloned from gist.github.com (detected by their git remote) and classifies
every URL: feeds, GitHub API endpoints, users, repos, gists, URL patterns, sites.

```python
from gitrecon.config import Config, load_env

load_env()  # GH_TOKEN / HF_TOKEN from dotenv files, never overriding your shell
config = Config()
```

```python
from collections import Counter

from gitrecon.config import PROJECT_ROOT
from gitrecon.sources.links import expand_pattern, gist_clones, harvest_gists

gists_root = PROJECT_ROOT.parent  # the scope directory holding the gist clones
print([clone.name for clone in gist_clones(gists_root)])
links = harvest_gists(gists_root)
Counter(link.kind for link in links)
```

```python
for link in links:
    if link.kind in ("feed", "github-api", "pattern"):
        print(f"{link.kind:<11} {link.url:<55} {link.found_in[0]}")
```

Patterns hold placeholders to fill in:

```python
expand_pattern("https://gist.github.com/{USER}.atom", user="sarverott")
```

Saving the catalog makes it the input of `feeds` and of the map update
(CLI: `gitrecon links --save`):

```python
from gitrecon.sources.links import save_catalog

save_catalog(links, config.data_dir / "catalog" / "links.jsonl")
```
