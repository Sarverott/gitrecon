# Data sources 3/5: the RFC index

The RFC Editor publishes one text file listing every RFC. `parse_index()` turns it into
`RFC` objects with title, authors, date, status, formats, DOI and relations
(obsoletes, updates, ...). It handles today's format: unpadded numbers and RFCs above 9999.

```python
from collections import Counter

from gitrecon.sources import rfc_index

rfcs = rfc_index.parse_index(rfc_index.fetch_index())
by_number = {rfc.number: rfc for rfc in rfcs}
len(rfcs), Counter(rfc.status for rfc in rfcs).most_common(5)
```

```python
process = by_number[2026]
process.title, process.status, process.obsoletes, process.updated_by[:5], process.url
```

Search like `gitrecon rfc --search ...`:

```python
words = ["quic"]
for rfc in rfcs:
    if all(w in rfc.description.lower() for w in words):
        print(f"RFC{rfc.number:<6} {rfc.date:<15} {rfc.title[:80]}")
```
