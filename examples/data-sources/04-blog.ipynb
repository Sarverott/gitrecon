# Data sources 4/4: blogs

Article links from a blog's index page, articles converted to markdown, and the feeds a
page advertises in its `<link rel="alternate">` tags (which the feed reader can then follow).

```python
from gitrecon.sources import blog

index_html = blog.fetch_html(blog.APOKRYF_BLOG_URL)
links = blog.article_links(index_html, blog.APOKRYF_BLOG_URL)
feeds = blog.discover_feeds(index_html, blog.APOKRYF_BLOG_URL)
links, feeds
```

```python
article = blog.html_to_article(links[0], blog.fetch_html(links[0]), blog.APOKRYF_BLOG_URL)
article.name, article.url_checksum, len(article.markdown)
```

```python
from IPython.display import Markdown

Markdown(article.to_markdown()[:3000])
```
