# The map dataset 3/3: pushing to the Hub

One `push()` uploads the whole local map as a single commit - or, with
`create_pr=True`, as a pull request on the Hub to review first.

**This writes to a public dataset.** The cell below only pushes when `PUSH = True`.
CLI: `gitrecon atlas push -m "..." [--pr]`, or `task map:push -- -m "..."`.

```python
from gitrecon.config import Config, load_env

load_env()  # GH_TOKEN / HF_TOKEN from dotenv files, never overriding your shell
config = Config()
```

```python
from gitrecon.hub.huggingface import MapDataset, default_local_dir

dataset = MapDataset(local_dir=default_local_dir(config))
PUSH = False
MESSAGE = "gitrecon: refresh map"

if PUSH:
    commit = dataset.push(MESSAGE, create_pr=False)
    print(commit)
else:
    print(f"dry run: would push {len(dataset.files())} files from {dataset.local_dir} to {dataset.repo_id}")
```

What is on the Hub now:

```python
from collections import Counter

from huggingface_hub import HfApi

api = HfApi()
remote = api.list_repo_files(dataset.repo_id, repo_type="dataset")
commits = api.list_repo_commits(dataset.repo_id, repo_type="dataset")
Counter(f.split("/")[0] for f in remote).most_common(), [c.title for c in commits[:3]]
```
