# The map dataset 1/3: getting the map

The map lives on the Hugging Face Hub as the dataset **Apokryf/minimap-of-uce** and locally
in `datasets/imperialmap` (git-ignored). `MapDataset` pulls and pushes it.

Credentials: `HF_TOKEN` from the environment or a dotenv file (`load_env()`), or an
interactive `login()`.

````python
from gitrecon.config import Config, load_env

load_env()  # GH_TOKEN / HF_TOKEN from dotenv files, never overriding your shell
config = Config()
````

````python
from gitrecon.hub.huggingface import DEFAULT_REPO_ID, MapDataset, default_local_dir

dataset = MapDataset(repo_id=DEFAULT_REPO_ID, local_dir=default_local_dir(config))
dataset.local_dir
````

Pull only when there is no local copy yet - pulling over local edits (a README drafted by
hand, say) is never what you want. Use `dataset.pull()` deliberately to refresh.

````python
if not (dataset.local_dir / "README.md").exists():
    dataset.pull()
dataset.areas()
````

````python
for area in dataset.areas():
    count = sum(1 for f in dataset.files() if f.relative_to(dataset.local_dir).parts[0] == area)
    print(f"{area:<22} {count:>6} files")
````

Areas are plain folders; a new one is created on demand:

````python
dataset.ensure_area("store-areas")
````

Without a token in the environment, log in interactively (stores the token for
`huggingface_hub`):

```python
from gitrecon.hub.huggingface import login
login()
```
