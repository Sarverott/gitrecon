# Text experiments: markdown, word-chains, numeronyms and RAT scripts

The first experiments of this project (formerly
`docs/tests-with-md-parsing-and-rattish-implementations.ipynb`), now on top of
`gitrecon.text`. Input: an LLM briefing about the Task tool.

```python
from IPython.display import HTML

from gitrecon.text import markdown

md_text = markdown.fetch_text(markdown.EXAMPLE_MD_URL)
html = markdown.md_to_html(md_text)
raw_text = markdown.html_to_text(html)
len(md_text), len(html), len(raw_text)
```

```python
HTML(html)
```

## Word-chains: symbols spelled out as tokens

Two spelling tables: the first, English one (`sentence_prep_legacy`) and the current
Polish one (`sentence_prep`, digits spelled too).

```python
from gitrecon.text.tokenizer import sentence_prep, sentence_prep_legacy, text_to_tokenchain_sentences

line = "Run: task build -- --force (v3.0)"
print(sentence_prep_legacy(line))
print(sentence_prep(line))
```

```python
sentences = text_to_tokenchain_sentences(raw_text)
len(sentences), sentences[:3]
```

## Numeronyms (`a12y`) with a collision glossary

`workflows` → `w9s`; words that collide get an index (`cross` → `c5s`, `codes` → `c5s1`).
The same scheme names directories in the map's `user-namespace`.

```python
from gitrecon.text.a12y import Glossary, paratangle

glossary = Glossary()
[glossary.a12y(w) for w in ["cross", "codes", "workflows", "_private"]], glossary.expand("c5s1"), paratangle("task")
```

## RAT script

The glossary section plus a sentence-rebuilding section. Output is byte-identical to the
original notebook for the same input (pinned in `tests/fixtures/taskfile-llms.*`).

```python
from gitrecon.text.rat import build_rat

build = build_rat(raw_text)
print("RAT size:", len(build.script), "lines:", build.lines, "glossary entries:", len(build.glossary))
print(build.script[:600])
```

## A JSON grammar in Lark

Raw API responses as a parse tree - for exploring structure, not for bulk data.

```python
from gitrecon.text.jsonlark import parse_json_tree

tree = parse_json_tree(markdown.fetch_text("https://api.github.com/gists/public?per_page=1"))
print(tree.pretty()[:1500])
```
